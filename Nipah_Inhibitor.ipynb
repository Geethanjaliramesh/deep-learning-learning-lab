```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/nipah_data_set/molecular_descriptor_training.csv")

print(df.shape)
print(df.head)
```

```python
print(df.columns.tolist())
```

```python
y = df['PIC50']
```

```python
x = df.drop(columns=['Name','IC50','PIC50','Unnamed: 45'])
```

```python
print('original dataset', df.shape)
print("x shape:", x.shape)
print('y shape:',y.shape)

print("\nFeature columns:")
print(x.columns.tolist())
```

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.dummy import DummyRegressor

```

```python
#5fold creation

cv = KFold(
    n_splits=5,
    shuffle = True,
    random_state=42
)

```

```python
baseline = DummyRegressor(strategy="mean")
```

```python
score = cross_validate(
    baseline,
    x,
    y,
    cv=cv,
    scoring=[
        "neg_mean_absolute_error",
        "neg_root_mean_squared_error",
        "r2"
    ]
)
```

```python
print(score)
```

```python
mae = -score["test_neg_mean_absolute_error"]
rmse = -score["test_neg_root_mean_squared_error"]
r2 = score["test_r2"]
print(f"Mean MAE:  {mae.mean():.3f}")
print(f"Mean RMSE: {rmse.mean():.3f}")
print(f"Mean R²:   {r2.mean():.3f}")
```

Dummy Mean Baseline — 5-Fold CV

MAE  ≈ 0.939
RMSE ≈ 1.139
R²   ≈ -0.195


Our dummy baseline's predictions are wrong by about 0.94 pIC50 units on average.

Now we are going to build our first model Linear reggression
Not because we expect it to be the best model. It's because it gives us another simple benchmark


```python
from sklearn.linear_model import LinearRegression
```

```python
linear_model = LinearRegression()
linear_scores = cross_validate(
    linear_model,
    x,
    y,
    cv=cv,
    scoring=[
        "neg_mean_absolute_error",
        "neg_root_mean_squared_error",
        "r2"
    ]
)
```

```python
linear_mae = -linear_scores["test_neg_mean_absolute_error"]
linear_rmse = -linear_scores["test_neg_root_mean_squared_error"]
linear_r2 = linear_scores["test_r2"]

print("Linear Regression Results")

print(f"Mean MAE:  {linear_mae.mean():.3f}")
print(f"Mean RMSE: {linear_rmse.mean():.3f}")
print(f"Mean R²:   {linear_r2.mean():.3f}")

print("\nMAE per fold:", linear_mae)
print("RMSE per fold:", linear_rmse)
print("R² per fold:", linear_r2)
```

# Scikit learn Pipeline
now we are intoducing the pipeline also we are standardising the features


```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
```

```python
ridge_model = Pipeline([
    ("scalar", StandardScaler()),
    ("ridge", Ridge(alpha=1.0))
])
```

```python
ridge_scores = cross_validate(
    ridge_model,
    x,
    y,
    cv=cv,
    scoring=[
        "neg_mean_absolute_error",
        "neg_root_mean_squared_error",
        "r2"]
)
```

```python
ridge_mae = -ridge_scores["test_neg_mean_absolute_error"]
ridge_rmse = -ridge_scores["test_neg_root_mean_squared_error"]
ridge_r2 = ridge_scores["test_r2"]

print("Ridge Regression Results")

print(f"Mean MAE:  {ridge_mae.mean():.3f}")
print(f"Mean RMSE: {ridge_rmse.mean():.3f}")
print(f"Mean R²:   {ridge_r2.mean():.3f}")

print("\nMAE per fold:", ridge_mae)
print("RMSE per fold:", ridge_rmse)
print("R² per fold:", ridge_r2)
```
